```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"E:\data_cleaning\customer_churn_sample.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Display the first 5 rows
df.head()
df.tail()
```

```python
print("Dataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Remove unnecessary spaces
# Convert column names to lowercase
# Replace spaces with underscores

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(df.columns.tolist())
```

```python
# Count missing values in every column
missing_values = df.isnull().sum()

print("Missing values:")
print(missing_values)
```

```python
categorical_columns = [
    "gender",
    "subscriptiontype",
    "contracttype",
    "paymentmethod",
    "churn"
]
```

```python
# Remove leading/trailing spaces
for column in categorical_columns:
    df[column] = df[column].astype(str).str.strip()

print("Categorical spaces cleaned successfully!")
```

```python
df["gender"] = (
    df["gender"]
    .str.strip()
    .str.lower()
    .str.capitalize()
)

print(df["gender"].unique())
```

```python
df["subscriptiontype"] = (
    df["subscriptiontype"]
    .str.strip()
    .str.title()
)

print(df["subscriptiontype"].unique())
```

```python
df["contracttype"] = df["contracttype"].str.strip()

df["paymentmethod"] = df["paymentmethod"].str.strip()

df["churn"] = (
    df["churn"]
    .str.strip()
    .str.capitalize()
)

print("Contract types:")
print(df["contracttype"].unique())

print("\nPayment methods:")
print(df["paymentmethod"].unique())

print("\nChurn values:")
print(df["churn"].unique())
```

```python
numeric_columns = [
    "age",
    "tenuremonths",
    "monthlycharges",
    "totalcharges",
    "supporttickets"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

print("Numeric columns converted successfully!")
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("==========================================")
print("       FINAL DATA VALIDATION")
print("==========================================")

# Dataset size
print("\n1. Dataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# Missing values
print("\n2. Missing Values:")
print(df.isnull().sum())

# Duplicate rows
print("\n3. Duplicate Rows:")
print(df.duplicated().sum())

# Duplicate Customer IDs
print("\n4. Duplicate Customer IDs:")
print(df["customerid"].duplicated().sum())

# Data types
print("\n5. Data Types:")
print(df.dtypes)

# Column names
print("\n6. Column Names:")
print(df.columns.tolist())

print("\n==========================================")
print("       VALIDATION COMPLETED")
print("==========================================")
```
